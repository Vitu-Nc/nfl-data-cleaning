```python
import pandas as pd
nfl_clean = pd.read_csv("data/nfl_data_clean.csv", keep_default_na=False, na_values=[""], low_memory=False)
nfl_clean['Date'] = pd.to_datetime(nfl_clean['Date'])
print(nfl_clean.shape)
```

```python
for i, c in enumerate(nfl_clean.columns[24:80]):
    print(i + 24, c)
```

```python
keep_cols = [
    # when and where
    'Date', 'Season', 'GameID', 'Drive', 'qtr', 'HomeTeam', 'AwayTeam',
    # situation
    'down', 'ydstogo', 'yrdline100', 'posteam', 'DefensiveTeam',
    'PosTeamScore', 'DefTeamScore', 'ScoreDiff',
    # outcome
    'PlayType', 'Yards.Gained', 'FirstDown', 'Touchdown', 'sp',
    # passing
    'Passer', 'Receiver', 'PassOutcome', 'PassLength', 'PassLocation',
    'AirYards', 'YardsAfterCatch', 'InterceptionThrown', 'Interceptor', 'Sack',
    # rushing
    'Rusher', 'RunLocation', 'RunGap', 'Fumble',
    # penalties and kicks
    'PenaltyType', 'Accepted.Penalty', 'Penalty.Yards',
    'FieldGoalResult', 'FieldGoalDistance',
    # advanced metrics
    'EPA', 'WPA', 'Win_Prob'
]

nfl_pbi = nfl_clean[keep_cols]
nfl_pbi.to_csv("data/nfl_powerbi.csv", index=False)
print(nfl_pbi.shape)
```
